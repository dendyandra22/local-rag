````python
import re
def response_sql_cleaner(text):
    text = text.replace("```","")
    text = text.replace('\n',' ')
    text = text.replace(';','')
    text = re.sub(r'(?i)^\s*sql\s*', '', text).strip()

    text = re.sub(r'\s+', ' ', text)


    return text

poi = 'SELECT *\nFROM manga\nWHERE score > 8.8 AND chapters > 20;'
response_sql_cleaner(poi)
````

````python
def _metadata_lookup_sql(nlu_model, message):
    query_filter = ''
    message = re.sub(
        r'\b(?:by\s+author|with\s+author|have\s+author|written\s+by|created\s+by|illustrated\s+by)\b',
        'by',
        message,
        flags=re.I
    )
    m = re.search(
        r'(.+?)\s+by\s+(.+)',
        message,
        flags=re.I
    )
    possible_title = []
    possible_authors = []

    if m:
        # print(m.group(1))
        possible_title = [f'{ent[1]}' for ent in nlu_model.predict_ner(m.group(1)) if ent[0] == 'MANGA_ENT']
        tmp_string = 'by ' + m.group(2)
        # print(tmp_string)
        possible_authors = [f'{ent[1]}' for ent in nlu_model.predict_ner(tmp_string) if ent[0] == 'MANGA_ENT']

    else:
        possible_title = nlu_model.predict_ner(message)

    # print(possible_title)
    # # print(m)
    # print(possible_authors)

    if possible_title:
        query_filter = query_filter + f'''\'(title:"{possible_title[0]}"'''
        query_filter = query_filter + f''' OR title_english:"{possible_title[0]}")'''

    if possible_authors:
        query_filter = query_filter + f''' AND authors:"{possible_authors[0]}"'''

    query_filter = query_filter + '\' ORDER BY score DESC'

    return query_filter
````

````python
def _entity_extraction(message, nlu_model):
    allowed_entities = ['MANGA_ENT', 'GENRES', 'PERIOD','RATING']
    message = re.sub(
            r'\b(?:by\s+author|with\s+author|have\s+author|written\s+by|created\s+by|illustrated\s+by)\b',
            'by',
            message,
            flags=re.I
    )
    m = re.search(
        r'(.+?)\s+by\s+(.+)',
        message,
        flags=re.I
    )

    if m:
        # print(m.group(1))
        tmp1 = [ent for ent in nlu_model.predict_ner(m.group(1)) if ent[0] in allowed_entities]
        tmp_string = 'by ' + m.group(2)
        # print(tmp_string)
        tmp2 = [ent for ent in nlu_model.predict_ner(tmp_string) if ent[0] in allowed_entities]
        tmp2 = [
            ('AUTHORS', ent[1]) if ent[0] == 'MANGA_ENT' else ent
            for ent in tmp2
        ]
        entities = tmp1 + tmp2

    else:
        entities = [ent for ent in nlu_model.predict_ner(message) if ent[0] in allowed_entities]

    return entities
````
